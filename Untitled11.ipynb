```python
# Step 1: Import Required Libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    roc_auc_score
)

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.callbacks import EarlyStopping

# Set random seeds for reproducibility
np.random.seed(42)
tf.random.set_seed(42)

print("TensorFlow version:", tf.__version__)
print("All libraries imported successfully!")
```

```python
# Step 2: Generate Dataset

X, y = make_moons(
    n_samples=1000,
    noise=0.20,
    random_state=42
)

print("Dataset generated successfully!")
print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("Number of classes:", len(np.unique(y)))
print("Class labels:", np.unique(y))

print("\nFirst 5 feature samples:")
print(X[:5])

print("\nFirst 5 target labels:")
print(y[:5])
```

```python
# Step 3: Visualize Dataset

plt.figure(figsize=(8, 6))

plt.scatter(
    X[:, 0],
    X[:, 1],
    c=y,
    cmap="coolwarm",
    edgecolors="black",
    s=40
)

plt.title("make_moons Dataset for Binary Classification")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.colorbar(label="Class")
plt.grid(alpha=0.3)
plt.show()
```

```python
# Step 4: Split Dataset and Standardize Features

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Standardize the features using training data only
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
print("Training labels:", y_train.shape)
print("Testing labels:", y_test.shape)
print("\nData preprocessing completed successfully!")
```

```python
# Step 5: Build ANN Model

model = Sequential([
    Input(shape=(2,)),
    Dense(16, activation="relu"),
    Dense(16, activation="relu"),
    Dense(1, activation="sigmoid")
])

# Compile the model
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# Display model architecture
model.summary()
```

```python
# Step 6: Train ANN

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True
)

history = model.fit(
    X_train,
    y_train,
    epochs=300,
    batch_size=32,
    validation_split=0.20,
    callbacks=[early_stopping],
    verbose=1
)

print("\nANN training completed successfully!")
```

```python
# Step 7: Evaluate ANN Model

test_loss, test_accuracy = model.evaluate(
    X_test,
    y_test,
    verbose=0
)

# Generate probabilities and predicted classes
y_prob = model.predict(X_test, verbose=0).ravel()
y_pred = (y_prob >= 0.5).astype(int)

# Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, y_prob)

print("=" * 45)
print("ANN MODEL EVALUATION RESULTS")
print("=" * 45)

print(f"Test Loss     : {test_loss:.4f}")
print(f"Test Accuracy : {test_accuracy * 100:.2f}%")
print(f"Accuracy Score: {accuracy * 100:.2f}%")
print(f"ROC-AUC Score : {roc_auc:.4f}")

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        target_names=["Class 0", "Class 1"],
        zero_division=0
    )
)
```

```python
# Step 8: Plot Accuracy

plt.figure(figsize=(8, 5))

plt.plot(history.history["accuracy"], label="Training Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")

plt.title("ANN Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
# Step 9: Plot Loss

plt.figure(figsize=(8, 5))

plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")

plt.title("ANN Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Binary Cross-Entropy Loss")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
# Step 10: Confusion Matrix

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Class 0", "Class 1"],
    yticklabels=["Class 0", "Class 1"]
)

plt.title("Confusion Matrix - ANN Binary Classification")
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.show()
```

```python
# Step 11: Plot Decision Boundary

# Create a mesh grid in the original feature space
x_min, x_max = X[:, 0].min() - 0.5, X[:, 0].max() + 0.5
y_min, y_max = X[:, 1].min() - 0.5, X[:, 1].max() + 0.5

xx, yy = np.meshgrid(
    np.linspace(x_min, x_max, 300),
    np.linspace(y_min, y_max, 300)
)

# Combine grid coordinates
grid = np.c_[xx.ravel(), yy.ravel()]

# Apply the same scaling used during training
grid_scaled = scaler.transform(grid)

# Predict class probabilities
Z = model.predict(grid_scaled, verbose=0)
Z = Z.reshape(xx.shape)

# Plot decision regions and original dataset
plt.figure(figsize=(9, 7))

plt.contourf(
    xx,
    yy,
    Z,
    levels=[0, 0.5, 1],
    cmap="coolwarm",
    alpha=0.35
)

plt.contour(
    xx,
    yy,
    Z,
    levels=[0.5],
    colors="black",
    linewidths=2
)

plt.scatter(
    X[:, 0],
    X[:, 1],
    c=y,
    cmap="coolwarm",
    edgecolors="black",
    s=35
)

plt.title("ANN Decision Boundary on make_moons Dataset")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.grid(alpha=0.2)
plt.show()
```

```python
# Step 12: Predict New Samples

new_samples = np.array([
    [-0.5, 0.5],
    [1.5, -0.5],
    [0.0, 1.0],
    [2.0, 0.0]
])

# Standardize new samples
new_samples_scaled = scaler.transform(new_samples)

# Predict probabilities
predicted_probabilities = model.predict(
    new_samples_scaled,
    verbose=0
).ravel()

# Convert probabilities into binary labels
predicted_classes = (
    predicted_probabilities >= 0.5
).astype(int)

print("NEW SAMPLE PREDICTIONS")
print("-" * 50)

for i in range(len(new_samples)):
    print(f"Sample: {new_samples[i]}")
    print(f"Predicted Class: {predicted_classes[i]}")
    print(f"Probability of Class 1: {predicted_probabilities[i]:.4f}")
    print("-" * 50)
```

```python
# Step 13: Final Summary

print("=" * 55)
print("ARTIFICIAL NEURAL NETWORK - FINAL RESULTS")
print("=" * 55)

print("Dataset             : make_moons (Scikit-learn)")
print("Problem Type        : Binary Classification")
print("Total Samples       : 1000")
print("Training Samples    :", len(X_train))
print("Testing Samples     :", len(X_test))
print("Hidden Layers       : 2")
print("Neurons per Layer   : 16, 16")
print("Hidden Activation   : ReLU")
print("Output Activation   : Sigmoid")
print("Optimizer           : Adam")
print("Loss Function       : Binary Cross-Entropy")
print("Epochs Completed    :", len(history.history["loss"]))
print(f"Test Accuracy       : {test_accuracy * 100:.2f}%")
print(f"Test Loss           : {test_loss:.4f}")
print(f"ROC-AUC Score       : {roc_auc:.4f}")

print("=" * 55)
print("ANN Binary Classification Completed Successfully!")
```
